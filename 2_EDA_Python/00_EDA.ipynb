```sql
%python
# cuales son los montos maximos minimo proemdio
df[['monto_facturado','monto_pagado']].describe().round(2)
```

```sql
%python
resumen_visita=(
    df.groupby("tipo_visita")
    .agg(trasaccion=("id_transaccion","count"),
         facturado=("monto_facturado","sum"),
         pagado=("monto_pagado","sum")
         )
    .reset_index()
    .sort_values("trasaccion",ascending=False)
    
)

display(resumen_visita)
```

```sql
%python
ax=sns.barplot(data=resumen_visita
            ,x="tipo_visita"
            ,y='facturado')
etiquetas = [f"{v/1_000_000:.1f}M".replace(".", ",") for v in ax.containers[0].datavalues]
ax.bar_label(ax.containers[0], labels=etiquetas)
       

ax.containers[0]
```

```sql
%python
# COMO CREAR HISTOGRAMAS
#case when 
df["rango_facturado"]=pd.cut(
    df['monto_facturado'],
    bins = [0,250,500,750,1000],
    labels = ['0-250','250-500','500-750','750-1000']

)
```

```sql
%python
display(df.groupby("rango_facturado",observed=True).size().reset_index(name='transacciones')
)
```
