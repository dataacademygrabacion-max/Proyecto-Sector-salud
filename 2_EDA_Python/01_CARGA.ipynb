```sql
%python
import pandas as pd
import os
```

```sql
%python
CATALOGO = "workspace"  # anlaogo a una base datos
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOGO}.salud_bronce")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOGO}.salud_plata")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOGO}.salud_oro")
spark.sql(f"CREATE VOLUME IF NOT EXISTS {CATALOGO}.salud_bronce.archivos")
```

```sql
%python
RUTA='/Workspace/Users/dataacademygrabacion@gmail.com/Proyecto-Sector-salud/data'
```

```sql
%python
#os.listdir(RUTA)
#archivo.replace('.csv','') 
# LIST CONPREMIDA 
TABLAS=[archivo.replace('.csv','') for archivo in os.listdir(RUTA) if archivo[-3:]=='csv']
# DICCIONARIO COMPRIMIDO
```

```sql
%python
# diccionrio comprimido
bronze={archivo.replace('.csv',''): pd.read_csv(f'{RUTA}/{archivo}') for archivo in os.listdir(RUTA) if archivo[-3:]=='csv'}
```

```sql
%python
for tabla in TABLAS:
    df=pd.read_csv(f'{RUTA}/{tabla}.csv')
    spark.createDataFrame(df).write.mode("overwrite").option("overwriteSchema","true")\
    .saveAsTable(f"{CATALOGO}.salud_bronce.{tabla}")
    print(f'Se cargo: {CATALOGO}.salud_bronce.{tabla}: {df.shape}')
```
